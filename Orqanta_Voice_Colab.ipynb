## 2. Initialize the Orqanta-Voice Engine

```python
import os
import sys
import soundfile as sf
from IPython.display import Audio
from orqanta_voice import OrqantaVoiceTTS

# Load Orqanta-Voice engine with 1 generation seat
tts = OrqantaVoiceTTS.from_pretrained(device="cuda", max_seats=1)
print("✅ Orqanta-Voice Engine Loaded Successfully!")
print("Available voices:", tts.voices)
```

## 3. Generate Speech in English & Indic Languages

```python
# 1. English Generation
text_en = "Hello! Welcome to Orqanta-Voice, streaming speech for 11 Indic languages and English."
wav_en = tts.generate(text_en, lang="en", voice="Siddhant")
sf.write("output_en.wav", wav_en, 24000)

# Play in Colab
Audio(wav_en, rate=24000)
```

```python
# 2. Hindi Generation (with numbers and currency)
text_hi = "नमस्ते! ओरकांता वॉइस में आपका स्वागत है। आज का तापमान 28 डिग्री है।"
wav_hi = tts.generate(text_hi, lang="hi", voice="Vaidehi")
sf.write("output_hi.wav", wav_hi, 24000)

# Play in Colab
Audio(wav_hi, rate=24000)
```

```python
# 3. Telugu Generation
text_te = "నమస్కారం! ఓర్కాంటా వాయిస్ కు స్వాగతం. మీరు ఎలా ఉన్నారు?"
wav_te = tts.generate(text_te, lang="te", voice="Sreekar")
sf.write("output_te.wav", wav_te, 24000)

# Play in Colab
Audio(wav_te, rate=24000)
```
